#### Setup Cell

Run the setup cell below to import the libraries and load the pre-split document chunks from the setup module. 

- `OpenAIEmbeddings` converts text into numerical vector representations using OpenAI's embedding model. 
- `FAISS` is a vector store built on Facebook AI Similarity Search that provides efficient similarity search over embedded documents. 
- `chunks` contains pre-split `Document` objects from `sample_document.txt`, loaded and split in `langchain_setup.py` using `TextLoader` and `RecursiveCharacterTextSplitter` from the previous exercise. 

**Note:** The OpenAI API key is already configured in this workspace, so no additional setup is needed. When working in an external environment, refer to the [OpenAI API](https://platform.openai.com/docs/quickstart?language=python) documentation for details on configuring your API key.

```python
# Third-party imports
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS

# Local application imports
from langchain_setup import chunks
```

**Note:** Be sure to run the Setup cell above before beginning with the checkpoints!

#### Checkpoint 1/4: Initializing the Embedding Model

Create an `OpenAIEmbeddings` instance named `embeddings` using `"text-embedding-3-small"` as the model to convert text chunks into numerical vectors that capture semantic meaning. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Initialize the OpenAI embedding model
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
```

#### Inspecting an Embedding

Run the code below to generate an embedding for a sample text and inspect its type, dimensions, and the first 10 values.

```python
# Generate an embedding for a sample text
sample_embedding = embeddings.embed_query("What is LangChain?")

print(f"Embedding type: {type(sample_embedding).__name__}")
print(f"Embedding dimensions: {len(sample_embedding)}")
print(f"First 10 values: {sample_embedding[:10]}")
```

#### Checkpoint 2/4: Creating a Vector Store from Chunks

Use `FAISS` to create a vector store named `vector_store` by passing `chunks` and `embeddings` as arguments to convert the document chunks into searchable embeddings stored in memory. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Create a FAISS vector store
vector_store = FAISS.from_documents(chunks, embeddings)
```

#### Inspecting the Vector Store

Run the code below to verify the vector store type and the number of embedded chunks stored and ready for similarity search. 

```python
# View vector store info
print(f"Vector store type: {type(vector_store).__name__}")
print(f"Number of vectors stored: {vector_store.index.ntotal}")
print(f"\nThe vector store now contains {vector_store.index.ntotal} embedded chunks,")
print(f"ready for similarity search.")
```

#### Checkpoint 3/4: Running a Similarity Search

With the vector store ready, we can now search it by meaning rather than by keywords. Use `similarity_search` on `vector_store` and store the result in a variable named `results`. 

- Pass `"What are the main components of LangChain?"` as the query to search for relevant chunks. 
- Set `k` to `3` to retrieve the top 3 most relevant chunks. 
- Uncomment the `print()` statements to display the query and its top results. 
   
**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Run a similarity search
results = vector_store.similarity_search(
    "What are the main components of LangChain?",
    k=3
)

# Display similarity search results
print(f"Query: \"What are the main components of LangChain?\"")
print(f"Top {len(results)} results:\n")

for i, doc in enumerate(results, 1):
    print(f"Result {i}")
    print("-" * 50)
    print(doc.page_content)
    print(f"Source: {doc.metadata}")
    print()
```

#### Checkpoint 4/4: Similarity Search with Scores

So far, search results have been returned without any indication of their relevance. `similarity_search_with_score` returns a relevance score for each result, indicating how closely each chunk matches the query. Store the result in a variable named `scored_results`. 

- Pass `"How does RAG work?"` as the query to search for relevant chunks. 
- Set `k` to `3` to retrieve the top 3 most relevant chunks. 
- Uncomment the `print()` statements to display the results and their scores. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Run a similarity search with scores
scored_results = vector_store.similarity_search_with_score(
    "How does RAG work?",
    k=3
)

# Display similarity search results with scores
print(f"Query: \"How does RAG work?\"")
print(f"Top {len(scored_results)} results with scores:\n")

for i, (doc, score) in enumerate(scored_results, 1):
    print(f"Result {i} (distance: {score:.4f})")
    print("-" * 50)
    print(doc.page_content)
    print()
```
