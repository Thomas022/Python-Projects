#### Setup Cell

Run the setup cell below to import the libraries required to load and split documents in LangChain. 

- `TextLoader`: A document loader that reads `.txt` files and converts them into LangChain Document objects. 
- `RecursiveCharacterTextSplitter`: A text splitter that breaks documents into smaller chunks using a hierarchy of separators (paragraphs, newlines, spaces). 

`sample_document.txt` is included in this workspace. It contains an overview of LangChain and its components. 

```python
# Third-party imports
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Path to the sample document included in this workspace
file_path = "sample_document.txt"
```

**Note:** Be sure to run the Setup cell above before beginning with the checkpoints!

#### Checkpoint 1/4: Loading a Text Document

Our first step is to load `sample_document.txt` into LangChain's Document format. 

- Create a `TextLoader` instance by passing `file_path` as the argument and store it in a variable named `loader` to prepare the file for reading. 
- Call `.load()` on `loader` to read the file and convert it into a list of `Document` objects, and store the result in a variable named `documents`. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Create a TextLoader instance for the file
loader = TextLoader(file_path)

# Load the file contents into Document objects
documents = loader.load()
```

#### Inspecting the Loaded Document

Run the block below to verify the document type, metadata, content length, and a preview of the first 100 characters.

```python
# View the loaded document
print(f"Number of documents loaded: {len(documents)}")
print(f"Document type: {type(documents[0]).__name__}")
print(f"\nMetadata: {documents[0].metadata}")
print(f"Content length: {len(documents[0].page_content)} characters")
print(f"\nFirst 100 characters:")
print(documents[0].page_content[:100])
```

#### Checkpoint 2/4: Creating a Text Splitter

Large documents need to be split into smaller chunks to fit within the model's context window and improve processing efficiency. Create a `RecursiveCharacterTextSplitter` instance named `text_splitter` with the following settings: 

- Set `chunk_size` to 200 to define the maximum number of characters per chunk. 
- Set `chunk_overlap` to 50 to retain overlapping characters between chunks and preserve context. 

`chunk_size` and `chunk_overlap` can be adjusted based on the document size and the level of context needed between chunks. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Create a text splitter for smaller chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=50
)
```

#### Checkpoint 3/4: Splitting the Document into Chunks

With the text splitter configured, we can now divide the document into smaller, processable segments that fit within the model's context window. 

Use the `split_documents` method on `text_splitter` to split documents into chunks and store the result in a variable named `chunks` to prepare them for downstream processing. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Split the documents into smaller chunks
chunks = text_splitter.split_documents(documents)
```

#### Exploring the Chunks

Run the code below to compare the original document length with the number of chunks produced and preview the content of each chunk.

```python
# Display chunk count and details
print(f"Original document: 1 document, {len(documents[0].page_content)} characters")
print(f"After splitting: {len(chunks)} chunks\n")

for i, chunk in enumerate(chunks, 1):
    print(f"CHUNK {i} ({len(chunk.page_content)} chars)")
    print("-" * 50)
    print(chunk.page_content)
    print()
```

#### Checkpoint 4/4: Experimenting with Chunk Parameters

A smaller `chunk_size` creates more, narrower chunks, while a larger one keeps more context in each chunk. Let's compare by creating a second splitter with larger settings. 

- Create a new `RecursiveCharacterTextSplitter` instance named `large_splitter` with `chunk_size` set to 500 and `chunk_overlap` set to 100 to allow larger chunks with more overlapping context. 
- Use `large_splitter` to split documents, storing the results in a variable named `large_chunks` to compare with the earlier chunks. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Create a text splitter for larger chunks
large_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

# Split the documents into larger chunks
large_chunks = large_splitter.split_documents(documents)
```

#### Comparing Chunk Sizes

Run the code below to compare the number and average length of chunks produced by each splitter and preview the content of the larger chunks. 

```python
# Compare the two chunking strategies
print("=" * 50)
print("COMPARISON: Small vs Large Chunks")
print("=" * 50)

print(f"\nSmall chunks (size=200, overlap=50):")
print(f"  Number of chunks: {len(chunks)}")
print(f"  Avg chunk length: {sum(len(c.page_content) for c in chunks) // len(chunks)} chars")

print(f"\nLarge chunks (size=500, overlap=100):")
print(f"  Number of chunks: {len(large_chunks)}")
print(f"  Avg chunk length: {sum(len(c.page_content) for c in large_chunks) // len(large_chunks)} chars")

print(f"\n--- Large Chunks Content ---")
for i, chunk in enumerate(large_chunks, 1):
    print(f"\nCHUNK {i} ({len(chunk.page_content)} chars)")
    print("-" * 50)
    print(chunk.page_content)
```
