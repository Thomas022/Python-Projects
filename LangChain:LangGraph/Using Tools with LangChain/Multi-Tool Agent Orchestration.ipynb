#### Setup Cell

Run the setup cell below to import the libraries required to build a custom agent toolkit in LangChain. 

- `AgentExecutor` runs the agent loop, managing tool calls and responses. 
- `create_tool_calling_agent` builds an agent that decides when and how to use tools. 
- `tool` converts Python functions into LangChain tools. 
- `BaseTool` and `BaseToolkit` provide the base classes for building custom tools and toolkits. 
- `ChatOpenAI` provides access to an OpenAI chat model. 
- `ChatPromptTemplate` and `MessagesPlaceholder` structure the prompt and track the agent's reasoning steps. 

The language model `llm` is initialized with `gpt-4o` at zero temperature for consistent, deterministic responses. 

**Note:** The OpenAI API key is already configured in this workspace, so no additional setup is needed. When working in an external environment, refer to the [OpenAI API](https://platform.openai.com/docs/quickstart?language=python) documentation for details on configuring your API key.

```python
from typing import List

# Core imports for agents and tools
from langchain.agents import AgentExecutor, create_tool_calling_agent
from langchain_core.tools import tool, BaseTool, BaseToolkit
from langchain_openai import ChatOpenAI

# For building prompts
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

# Initialize the language model
llm = ChatOpenAI(model="gpt-4o", temperature=0, stream=False)
```

**Note:** Be sure to run the Setup cell above before beginning with the checkpoints!

#### Checkpoint 1/5: Creating a Math Toolkit

A toolkit groups related tools into a single, reusable unit. Let's start by building `MathToolkit`, which bundles three arithmetic operations together. 

- Use the `@tool` decorator to define three functions: `add`, `subtract`, and `multiply`. Each function takes two `float` parameters, `a` and `b`, and includes a docstring that describes the operation and when to use it. 
- Create a class named `MathToolkit` that inherits from `BaseToolkit`:  
   - Add a docstring that describes the toolkit's purpose. 
   - Define a `get_tools()` method that returns `add`, `subtract`, and `multiply` as a list of `BaseTool` objects. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Define individual math tools

@tool
def add(a: float, b: float) -> float:
    """Add two numbers. Use for addition operations."""
    return a + b

@tool
def subtract(a: float, b: float) -> float:
    """Subtract two numbers. Use for subtraction operations."""
    return a - b

@tool
def multiply(a: float, b: float) -> float:
    """Multiply two numbers. Use for multiplication operations."""
    return a * b

# Create the MathToolkit class
class MathToolkit(BaseToolkit):
    """Toolkit for mathematical operations."""
    
    def get_tools(self) -> List[BaseTool]:
        """Return all math tools."""
        return [add, subtract, multiply]
```

#### Testing the Math Toolkit

Run the block below to verify that `MathToolkit` returns all three tools and that `add` and `multiply` produce the correct results. 

```python
# Test the MathToolkit
math_toolkit = MathToolkit()
math_tools = math_toolkit.get_tools()

print(f"MathToolkit contains {len(math_tools)} tools:")
for t in math_tools:
    print(f"  - {t.name}")

# Test individual operations
print(f"\nTest: 10 + 5 = {add.invoke({'a': 10, 'b': 5})}")
print(f"Test: 5 * 3 = {multiply.invoke({'a': 5, 'b': 3})}")
```

#### Checkpoint 2/5: Creating a String Toolkit

Now let's build `StringToolkit`, which bundles three string operations together following the same pattern as `MathToolkit`. 

- Use the `@tool` decorator to define three functions: `to_uppercase`, `count_words`, and `reverse_text`. Each function takes a single `str` parameter named `text` and includes a docstring that describes the operation and when to use it. 
- Create a class named `StringToolkit` that inherits from `BaseToolkit`:  
   - Add a docstring that describes the toolkit's purpose. 
   - Define a `get_tools()` method that returns `to_uppercase`, `count_words`, and `reverse_text` as a list of `BaseTool` objects. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Define individual string tools

@tool
def to_uppercase(text: str) -> str:
    """Convert text to uppercase. Use for uppercase conversion."""
    return text.upper()

@tool
def count_words(text: str) -> int:
    """Count words in text. Use for word counting."""
    return len(text.split())

@tool
def reverse_text(text: str) -> str:
    """Reverse text. Use for text reversal."""
    return "".join(reversed(text))

# Create the StringToolkit class
class StringToolkit(BaseToolkit):
    """Toolkit for string operations."""
    
    def get_tools(self) -> List[BaseTool]:
        """Return all string tools."""
        return [to_uppercase, count_words, reverse_text]
```

#### Testing the String Toolkit

Run the block below to verify that `StringToolkit` returns all three tools and that `to_uppercase`, `count_words`, and `reverse_text` produce the correct results. 

```python
# Test the StringToolkit
string_toolkit = StringToolkit()
string_tools = string_toolkit.get_tools()

print(f"StringToolkit contains {len(string_tools)} tools:")
for t in string_tools:
    print(f"  - {t.name}")

# Test individual operations
sample = "Hello World"
print(f"\nTest text: '{sample}'")
print(f"  uppercase: {to_uppercase.invoke({'text': sample})}")
print(f"  word count: {count_words.invoke({'text': sample})}")
print(f"  reversed: {reverse_text.invoke({'text': sample})}")
```

#### Checkpoint 3/5: Creating Toolkit Instances and Combining Tools

With both toolkits defined, we can now create instances of each and combine their tools into a single list that the agent can use. 

- Create an instance of `MathToolkit` and store it in a variable named `math_toolkit` to access its arithmetic tools. 
- Create an instance of `StringToolkit` and store it in a variable named `string_toolkit` to access its string tools. 
- Call `get_tools()` on each toolkit and concatenate the results into a single list named `tools` so the agent has access to all available tools. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Create toolkit instances
math_toolkit = MathToolkit()
string_toolkit = StringToolkit()

# Combine all tools from both toolkits
tools = math_toolkit.get_tools() + string_toolkit.get_tools()
```

#### Checkpoint 4/5: Building the Multi-Tool Agent

Now that the tools are combined into a single list, let's build the agent and wire everything together into a working system. 

- Create a prompt template named `agent_prompt` with three components to define how the agent communicates and reasons: 
   - A system message that describes the assistant as helpful and clearly instructs it to use available tools when answering questions 
   - A human message with an `input` placeholder for the user's question 
   - A `MessagesPlaceholder` with `variable_name` set to `"agent_scratchpad"` to store the agent's reasoning steps and tool results 
- Use `create_tool_calling_agent` to create an agent named `agent`, passing in `llm`, `tools`, and `agent_prompt` so the agent knows which model, tools, and prompt to use. 
- Instantiate `AgentExecutor` and store it in a variable named `agent_executor` to manage the agent's run loop:  
   - Pass in `agent` and `tools`. 
   - Set `verbose` to `True` to display the agent's reasoning during execution. 
   - Set `stream_runnable` to `False` to disable streaming. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Create the prompt template
agent_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Use your tools to answer questions."),
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),
])

# Create the agent
agent = create_tool_calling_agent(llm, tools, agent_prompt)

# Create the executor
agent_executor = AgentExecutor(
    agent=agent,
    tools=tools,
    verbose=True,
    stream_runnable=False
)
```

#### Checkpoint 5/5: Testing Tool Selection

With the agent fully assembled, let's test it across five questions to verify that it selects the correct tool for each task. 

- Invoke `agent_executor` for each question in `test_questions` and store the result in a variable named `result`. 
- Uncomment the `print()` statements to display each question and its corresponding answer. 

**Note:** Don't forget to run the cell and save the notebook before selecting `Test Work`. Open the `Jupyter Help` toggle at the top of the notebook for more details.

```python
# Test questions that require different toolkits
test_questions = [
    "I bought 4 boxes with 25 items in each. How many items do I have in total?",
    "I need this text in all caps for a heading: 'hello world'.",
    "Can you tell me how many words are there in the sentence 'The quick brown fox jumps'?",
    "I started with 100 points and spent 40. How many points do I have left?",
    "For a palindrome check, can you reverse the word 'LangChain'?",
]

for question in test_questions:
    result = agent_executor.invoke({"input": question})
    print(f"\nQuestion: {question}")
    print(f"Answer: {result['output']}")
    print("-" * 50)
```
